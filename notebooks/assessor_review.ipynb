```python
import pandas as pd
import numpy as np
```

```python
assessor1 = pd.read_excel('/Users/ishanmaharjan/WorkSpace/WORK/AI chatbot for crisis support/data/assessor_replies/Assessor 1.xlsx')
assessor2 = pd.read_excel('/Users/ishanmaharjan/WorkSpace/WORK/AI chatbot for crisis support/data/assessor_replies/Assessor 2.xlsx')
```

```python
assessor1['assessor_judgment'].value_counts()
```

```python
assessor2['assessor_judgment'].value_counts()
```

```python
assessor1['assessor_judgment'].isnull().sum()
```

```python
assessor2['assessor_judgment'].isnull().sum()
```

```python
assessor1["assessor_judgment"] = assessor1["assessor_judgment"].replace({"YES": "Yes"})
assessor1["assessor_judgment"] = assessor1["assessor_judgment"].replace({"NO": "No"})
```

```python
assessor1["assessor_judgment"].value_counts()
```

```python
assessor2["assessor_judgment"] = assessor2["assessor_judgment"].replace({"yes": "Yes"})
assessor2["assessor_judgment"] = assessor2["assessor_judgment"].replace({"NO": "No"})
```

```python
assessor2["assessor_judgment"].value_counts()
```

```python
assessor1["assessor_judgment"] = assessor1["assessor_judgment"].fillna("NA")

assessor1["assessor_judgment"].isnull().sum()
```

```python
assessor1["assessor_judgment"].value_counts()
```

```python
assessor2["assessor_judgment"] = assessor2["assessor_judgment"].fillna("NA")
assessor2["assessor_judgment"] = assessor2["assessor_judgment"].replace({"na": "NA"})
```

```python
assessor2["assessor_judgment"].value_counts()
```

```python
assessor1.to_excel("cleaned assessor 1.xlsx", index=False)
assessor2.to_excel("cleaned assessor 2.xlsx", index=False)
```
